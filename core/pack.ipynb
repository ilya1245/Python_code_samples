```python
user_record = ('Андрей', 'andrew@example.com', '25-333-26', '888-12-11')
name, email, *phone_numbers = user_record
print(name, email)
print(phone_numbers)
```

```python
*some_data, phone_1, phone_2 = user_record
print(some_data)
print(phone_1, phone_2)
```

```python
a, *b, c = user_record
print(a)
print(b)
print(c)
```
